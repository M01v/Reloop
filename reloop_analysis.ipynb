# Support chat

```python
chat_text = Path("support_chat.txt").read_text(encoding="utf-8")
print(chat_text[:3000])
```

```python
import re

sections = re.split( r"===\s*(D\d+)\s*===",chat_text)

print("Number of dispute sections:", (len(sections) - 1) // 2)

print("\nFirst dispute ID:", sections[1])
print("\nFirst dispute conversation:")
print(sections[2].strip())
```

```python
message_pattern = (
    r"\[(\d{4}-\d{2}-\d{2})\]\s*"
    r"(Buyer|Seller):\s*(.*)"
)

first_chat = sections[2].strip()

messages = re.findall(message_pattern, first_chat)

for date, speaker, message in messages:
    print({
        "date": date,
        "speaker": speaker,
        "message": message
    })
```

# Combine the three data sources

```python
# Check how the data connects
print("Orders columns:")
print(orders.columns.tolist())

print("\nTracking columns:")
print(tracking.columns.tolist())

print("\nFirst parsed dispute IDs:")
print(sections[1:10:2])
```

```python
# parsed messages into a DataFrame
chat_records = []

for i in range(1, len(sections), 2):
    dispute_id = sections[i].strip()
    conversation = sections[i + 1].strip()

    messages = re.findall(
        r"\[(\d{4}-\d{2}-\d{2})\]\s*"
        r"(Buyer|Seller):\s*(.*)",
        conversation
    )

    for date, speaker, message in messages:
        chat_records.append({
            "dispute_id": dispute_id,
            "date": date,
            "speaker": speaker,
            "message": message
        })

chat_df = pd.DataFrame(chat_records)

display(chat_df.head(10))
print("Total messages:", len(chat_df))
print("Total disputes:", chat_df["dispute_id"].nunique())
```

```python
# one conversation summary per dispute
chat_summary = (
    chat_df.sort_values("date")
    .groupby("dispute_id")
    .apply(
        lambda group: "\n".join(
            f"[{row.date}] {row.speaker}: {row.message}"
            for row in group.itertuples()
        )
    )
    .reset_index(name="conversation")
)

display(chat_summary.head())
#chat_summary
```

```python
# Join the data
evidence = orders.merge(
    tracking,
    on="order_id",
    how="left",
    validate="one_to_one",
    indicator=True
)

print("Evidence shape:", evidence.shape)
print("\nTracking match status:")
print(evidence["_merge"].value_counts())

display(evidence.head()) 
# combines the two datasets using their shared order_id.
```

```python
print("Order IDs:")
print(orders["order_id"].tolist())

print("\nDispute IDs:")
print(chat_summary["dispute_id"].tolist())
```

# complete evidence table

```python
# Merge orders with tracking
evidence = orders.merge(
    tracking,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Merge the support conversations
evidence = evidence.merge(
    chat_summary,
    left_on="order_id",
    right_on="dispute_id",
    how="left",
    validate="one_to_one"
)

# Remove the redundant ID column
evidence = evidence.drop(columns=["dispute_id"])

print("Total disputes:", len(evidence))
print("Missing tracking records:", evidence["delivery_status"].isna().sum())
print("Missing conversations:", evidence["conversation"].isna().sum())

display(evidence.head())
```

# Decision Engine

```python

def clean_text(value):
    """Convert missing values to empty strings and normalize text."""
    if pd.isna(value):
        return ""
    return str(value).lower().strip()


def has_any(text, phrases):
    return any(phrase in text for phrase in phrases)


def make_decision(row):
    # -----------------------------
    # 1. Read and normalize evidence
    # -----------------------------
    reason = clean_text(row.get("dispute_reason", ""))
    status = clean_text(row.get("delivery_status", ""))
    notes = clean_text(row.get("delivery_notes", ""))
    signed_by = clean_text(row.get("signed_by", ""))
    chat = clean_text(row.get("conversation", ""))

    evidence = (
        f"Reason: {reason}; Status: {status}; "
        f"Signed by: {signed_by or 'not recorded'}; "
        f"Tracking notes: {notes}; Conversation: {chat}"
    )

    def result(verdict, confidence, explanation, rule):
        return {
            "verdict": verdict,
            "confidence": confidence,
            "reason": explanation,
            "evidence": evidence,
            "rule_applied": rule
        }

    # -----------------------------
    # 2. Identify the dispute type
    # -----------------------------
    non_receipt = has_any(reason, [
        "never received",
        "not received",
        "item not received",
        "never arrived",
        "did not arrive"
    ])

    damage_issue = has_any(reason, [
        "damaged",
        "damage",
        "broken",
        "does not power on",
        "doesn't power on",
        "not working",
        "faulty",
        "defective"
    ])

    # -----------------------------
    # RULE 1: Non-delivery
    # -----------------------------
    non_delivery_status = has_any(status, [
        "in transit",
        "label created",
        "not picked up",
        "lost",
        "returned to sender",
        "exception"
    ])

    non_delivery_notes = has_any(notes, [
        "never picked up",
        "not yet delivered",
        "lost in transit",
        "never collected",
        "not handed to carrier"
    ])

    wrong_address_delivery = (
        "delivered to wrong address" in notes
        or "delivered to a different address" in notes
    )

    if (
        non_delivery_status
        or non_delivery_notes
        or wrong_address_delivery
    ):
        return result(
            "refund",
            "high",
            "Tracking evidence indicates the item was not delivered "
            "to the buyer as expected.",
            "Rule 1 — Non-delivery"
        )

    # -----------------------------
    # RULE 2: Seller acknowledges error
    # -----------------------------
    seller_error_phrases = [
        "mixed up your order",
        "mixed up the order",
        "sent the wrong item",
        "shipped the wrong item",
        "grabbed the wrong one",
        "wrong item",
        "mislabeled the box",
        "mislabelled the box",
        "wrong size",
        "incorrect size",
        "listing was incorrect",
        "listing was wrong",
        "my mistake",
        "that's on me",
        "that is on me"
    ]

    seller_admits_error = has_any(chat, seller_error_phrases)

    # A specific seller admission of a listing/fulfillment error
    # takes precedence over later generic delivery evidence.
    if seller_admits_error:
        return result(
            "refund",
            "high",
            "The seller acknowledges a listing or fulfillment error.",
            "Rule 2 — Seller acknowledges error"
        )

    # -----------------------------
    # Seller confirms damage coverage
    # -----------------------------
    damage_coverage_phrases = [
        "damage is covered",
        "doa damage is covered",
        "covered under our policy",
        "covered by our policy"
    ]

    refund_commitment_phrases = [
        "refunding you",
        "refunding now",
        "i'll refund",
        "i will refund",
        "i'll issue a refund",
        "i will issue a refund",
        "issuing a refund",
        "refund you"
    ]

    if (
        damage_issue
        and has_any(chat, damage_coverage_phrases)
        and has_any(chat, refund_commitment_phrases)
    ):
        return result(
            "refund",
            "high",
            "The seller confirms that the reported damage is covered "
            "and explicitly commits to a refund.",
            "Seller-confirmed damage coverage"
        )

    # A seller explicitly promising a refund for a reported fault
    if (
        damage_issue
        and has_any(chat, refund_commitment_phrases)
        and has_any(chat, [
            "that's a real fault",
            "that is a real fault",
            "packaging was too thin",
            "packaging was inadequate",
            "item was faulty",
            "item is faulty",
            "arrived broken",
            "doesn't work",
            "does not work"
        ])
    ):
        return result(
            "refund",
            "high",
            "The seller acknowledges the reported fault and explicitly "
            "promises a refund.",
            "Seller-confirmed fault and refund"
        )

    # -----------------------------
    # RULE 3 / 4: Non-receipt claims
    # -----------------------------
    buyer_receipt_phrases = [
        "i received it",
        "i got it",
        "it arrived",
        "package arrived",
        "item arrived",
        "got the package",
        "got the item",
        "love the jacket",
        "thanks, got it"
    ]

    buyer_denies_receipt_phrases = [
        "never got this",
        "never received",
        "i didn't receive",
        "i did not receive",
        "i never got",
        "someone else must have signed",
        "someone else signed",
        "not my signature",
        "i wasn't home",
        "i was not home"
    ]

    receipt_retraction_phrases = [
        "maybe it was a different package",
        "might have been a different package",
        "packages were mixed",
        "not sure that was this item",
        "i may have been mistaken"
    ]

    buyer_acknowledged_receipt = has_any(
        chat, buyer_receipt_phrases
    )

    buyer_denies_receipt = has_any(
        chat, buyer_denies_receipt_phrases
    )

    receipt_retracted = has_any(
        chat, receipt_retraction_phrases
    )

    signed_delivery = (
        status == "delivered"
        and signed_by not in ["", "nan", "none", "not recorded", "unknown"]
    )

    delivered = "delivered" in status

    # A buyer disputes who signed: evidence conflicts.
    if non_receipt and has_any(chat, [
        "someone else must have signed",
        "someone else signed",
        "not my signature"
    ]):
        return result(
            "escalate",
            "medium",
            "Tracking records a signature, but the buyer disputes "
            "who signed for the delivery.",
            "Rule 4 — Conflicting evidence"
        )

    # Receipt acknowledged, then contradicted or retracted.
    if (
        non_receipt
        and buyer_acknowledged_receipt
        and (buyer_denies_receipt or receipt_retracted)
    ):
        return result(
            "escalate",
            "low",
            "Receipt evidence is contradicted or retracted in the "
            "conversation.",
            "Rule 4 — Conflicting evidence"
        )

    # Credible receipt evidence conflicts with a non-receipt claim.
    if non_receipt and buyer_acknowledged_receipt:
        return result(
            "deny",
            "medium",
            "The buyer's own messages indicate receipt, which conflicts "
            "with the non-receipt claim.",
            "Rule 3 — Confirmed receipt"
        )

    # Signed delivery alone is not enough if the buyer disputes it.
    if non_receipt and signed_delivery:
        return result(
            "escalate",
            "medium",
            "Tracking records a signature, but the available evidence "
            "does not establish that the buyer received the item.",
            "Rule 4 — Conflicting or insufficient evidence"
        )

    # Delivered but no reliable receipt confirmation.
    if non_receipt and delivered:
        return result(
            "escalate",
            "low",
            "Tracking says delivered, but there is no reliable evidence "
            "that the buyer received the item.",
            "Rule 4 — Insufficient evidence"
        )

    # Non-receipt with unclear tracking.
    if non_receipt:
        return result(
            "escalate",
            "low",
            "The available tracking and conversation evidence is "
            "insufficient to determine whether delivery occurred.",
            "Rule 4 — Insufficient evidence"
        )

    # -----------------------------
    # RULE 5: Already resolved
    # -----------------------------
    buyer_acceptance_phrases = [
        "partial refund is fine",
        "i accept the partial refund",
        "i accept the refund",
        "that works for me",
        "exchange is fine",
        "i agree to the exchange",
        "i accept the exchange"
    ]

    seller_resolution_phrases = [
        "partial refund",
        "offer an exchange",
        "offer a refund",
        "i'll refund",
        "i will refund",
        "refunding now",
        "send a replacement",
        "replace the item"
    ]

    if (
        has_any(chat, buyer_acceptance_phrases)
        and has_any(chat, seller_resolution_phrases)
    ):
        return result(
            "deny",
            "medium",
            "The conversation indicates that both parties accepted "
            "a resolution to the issue.",
            "Rule 5 — Issue already resolved"
        )

    # -----------------------------
    # RULE 4: Default to escalation
    # -----------------------------
    return result(
        "escalate",
        "low",
        "Evidence is insufficient or conflicting, so the case requires "
        "manual review.",
        "Rule 4 — Insufficient or conflicting evidence"
    )


# --------------------------------
# Apply decisions to all disputes
# --------------------------------
decisions = evidence.apply(make_decision, axis=1, result_type="expand")

results = pd.concat(
    [
        evidence[
            [
                "order_id",
                "buyer_handle",
                "seller_handle",
                "dispute_reason"
            ]
        ].reset_index(drop=True),
        decisions.reset_index(drop=True)
    ],
    axis=1
)

# --------------------------------
# Add history review flags
# --------------------------------
buyer_counts = evidence["buyer_handle"].value_counts()
seller_counts = evidence["seller_handle"].value_counts()

results["repeat_buyer_flag"] = (
    results["buyer_handle"].map(buyer_counts).gt(1)
)

results["repeat_seller_flag"] = (
    results["seller_handle"].map(seller_counts).gt(1)
)

results["history_review"] = (
    results["repeat_buyer_flag"]
    | results["repeat_seller_flag"]
)

# --------------------------------
# Review and export
# --------------------------------
print("Verdict counts:")
print(results["verdict"].value_counts())

print("\nDecision preview:")
display(results)

results.to_csv("reloop_decisions.csv", index=False)

print("\nSaved: reloop_decisions.csv")
```
